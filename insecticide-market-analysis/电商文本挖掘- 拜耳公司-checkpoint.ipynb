```python
# 遍历7个市场的交易数据
df_list = [ ]

for file in files:
    
    # 读取文件内容
    df = pd.read_excel(file)
    # print(df.时间.dtype)  # 'int64'
    if df.时间.dtype == 'int64':
        # 把时间数字 转换成日期
        df.时间 = pd.to_datetime(df.时间, unit='D', origin='1899-12-30')
    
    # 将时间变成行索引
    df.set_index(keys='时间', inplace=True)
    # display(df.head())
    
    # 修改列名
    filename = file.split('市场近三年交易额')[0]
    df.columns = [filename]
    
    df_list.append(df)
```

```python
# 水平合并
df = pd.concat(df_list, axis=1)
df
```

```python
# 重置索引
df.reset_index(inplace=True)
```

###  线性回归： 预测2018年11月，12月的数据

```python
from sklearn.linear_model import LinearRegression
```

```python
# 先考虑11月的预测
X_train = np.array([2015, 2016, 2017]).reshape(-1, 1)
y_train = df.loc[df.时间.dt.month == 11,  '灭鼠杀虫剂'].values[: : -1]

X_test = np.array([2018]).reshape(-1, 1)

# 线性回归
linear = LinearRegression()
linear.fit(X_train, y_train)

# 预测
linear.predict(X_test)
```

#### 考虑11月，12月

```python
df.head(3)
```

```python
import datetime
```

```python

# 遍历11月，12月
for month in [11, 12]:
    
    y_list = [ datetime.datetime(year=2018, month=month, day=1) ]
    
    # 遍历每一个产品
    for col in df.columns[1: ]:
        
        # 训练数据
        X_train = np.array([2015, 2016, 2017]).reshape(-1, 1)
        y_train = df.loc[df.时间.dt.month == month,  col].values[: : -1]
        X_test = np.array([2018]).reshape(-1, 1)

        # 线性回归
        linear = LinearRegression()
        linear.fit(X_train, y_train)

        # 预测
        y = linear.predict(X_test)
        # print(month, col, y)
        y_list.append(y[0])
    
    # print(y_list)
    # 把y_list转换成DataFrame
    y_array = np.array(y_list).reshape(1, -1)
    # display(y_array)
    y_df = pd.DataFrame(data=y_array, columns=df.columns)
    display(y_df)
    
    # 拼接
    df = y_df.append(df)
```

### 进一步分析各类目市场16-17年17-18年增幅

```python

# 16-17年增幅 =  (2017年销售额 - 2016年销售额)  / 2016年销售额
# 17-18年增幅 =  (2018年销售额 - 2017年销售额)  / 2017年销售额
```

```python
year_df
```

```python
# # 16-17年增幅 =  (2017年销售额 - 2016年销售额)  / 2016年销售额
growth_16_17 = (year_df.loc[2017] - year_df.loc[2016] ) / year_df.loc[2016] 
growth_16_17
```

```python
# 17-18年增幅 =  (2018年销售额 - 2017年销售额)  / 2017年销售额
growth_17_18 = (year_df.loc[2018] - year_df.loc[2017] ) / year_df.loc[2017] 
growth_17_18
```

```python
year_df2 = pd.concat((growth_16_17, growth_17_18), axis=1)
year_df2.columns = ['16-17', '17-18']
year_df2
```

```python
year_df2.T
```

```python
year_df2.T.plot(figsize=(10, 6))

plt.xticks([0, 1], ['16-17', '17-18'])
plt.xlabel('年份')
plt.ylabel('总交易额年增幅')
plt.title('近三年总交易额年增幅')
plt.show()

# 可以看出：
#    灭鼠杀虫剂，蚊香加热器，防霉防蛀片增长幅度相对来说稍微稳定一点
#    电蚊香套装降幅最快， 蚊香液，蚊香片，盘香灭蟑香蚊香盘下降也很大
```
