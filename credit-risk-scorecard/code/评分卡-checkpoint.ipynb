```python
# 获取历史数据，定义观察时间窗口


sample_window = pd.read_csv('data/CreditSampleWindow.csv')
```

```python
# CID         用户ID
# STAGE_BEF   本阶段的逾期阶段 
# STAGE_AFT   本阶段进入的逾期阶段
# START_DATE  进入本阶段时间
# CLOSE_DATE  本阶段的结束时间

# 其中：逾期阶段含义为：
# M0: 逾期 0-3 天
# M1: 逾期 3-30 天
# M2: 逾期 30-60 天
# M3: 逾期 60-90 天
# ，以此类推 

sample_window.head()
```

```python
# 构建一个转移矩阵

def gen_trans_mat(df):
    # 10行 * 10列
    trans_mat = pd.DataFrame(
        data=np.zeros(shape=(10,10)),
        index=np.arange(10),
        columns=np.arange(10)
    )
    # print(trans_mat)
    
    counter = collections.Counter()
    
    # 遍历所有的情况
    for i, j in zip(df.columns[:-1], df.columns[1:]):
        # print(i, j)
        data = df[[i, j]]
        select = data.notnull().all(axis=1)
        # print(select)
        
        # print(data.loc[select].values)
        data2 = [tuple(x) for x in data.loc[select].values]
        # print(data2)
        counter += collections.Counter(data2)
        
    # print(counter)
    
    for index, column in counter.keys():
        # print(index, column)
        trans_mat.loc[index, column] = counter[(index, column)]
    
    # all_count
    trans_mat['all_count'] = trans_mat.sum(axis=1)
    
    # bad_count
    bad_counts = []
    for n in trans_mat.index:
        bad_count = trans_mat.loc[n][n+1:10].sum()
        bad_counts.append(bad_count)
    trans_mat['bad_count'] = bad_counts
    
    # to_bad
    trans_mat['to_bad'] = trans_mat['bad_count'] / trans_mat['all_count']
    
    # print(trans_mat)
    return trans_mat
    
# 调用
# gen_trans_mat(result)
```

```python
# 调用,构建一个转移矩阵
trans_mat = gen_trans_mat(result)
```

段时，下一阶段继续转坏的概率为 89%，可根据业务需要（营销、风险等等）来考虑定义进入 M2 阶段的用户为坏客户
